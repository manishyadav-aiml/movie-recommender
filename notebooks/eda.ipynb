# Movie Recommendation System - EDA & Evaluation
Run from the project root: `jupyter notebook notebooks/eda.ipynb`

```python
import sys; sys.path.insert(0, '..')
import pandas as pd, matplotlib.pyplot as plt
from src import MovieRecommender
rec = MovieRecommender('ml-1m')
movies, ratings = rec.movies, rec.ratings
print(movies.shape, ratings.shape)
movies.head()
```

## Ratings

```python
ratings['rating'].value_counts().sort_index().plot.bar(title='Rating distribution')
plt.show()
```

## Long tail
A few blockbusters get most ratings; most movies get few. This is why we add shrinkage and a popularity prior.

```python
movies['n_ratings'].sort_values(ascending=False).reset_index(drop=True).plot(logy=True, title='Ratings per movie')
plt.show()
```

## Genres

```python
pd.Series([g for gs in movies['genres'] for g in gs.split('|') if g]).value_counts().sort_values().plot.barh(figsize=(7,6))
plt.show()
```

## Recommendations

```python
rec.recommend('The Matrix', n=8)
```

```python
rec.recommend_for_ratings({'The Godfather': 5, 'Heat': 4.5}, n=8)
```
